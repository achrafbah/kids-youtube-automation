```python
# 6. عرض الفيديو الناتج مباشرة في المتصفح
import glob
from IPython.display import HTML
from base64 import b64encode

output_videos = glob.glob("animations/*_animated.mp4")
if output_videos:
    target_video = output_videos[-1]
    mp4 = open(target_video, 'rb').read()
    data_url = "data:video/mp4;base64," + b64encode(mp4).decode()
    display(HTML(f"""
    <video width=600 controls autoplay loop>
          <source src="{data_url}" type="video/mp4">
    </video>
    """))
else:
    print("تأكد من مسار الفيديو الناتج ف مجلد animations/")
```
